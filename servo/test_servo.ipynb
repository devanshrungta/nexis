```python
import Jetson.GPIO as GPIO
import time

servoPin = 32

def setup():
    GPIO.setmode(GPIO.BOARD)
    GPIO.setup(servoPin, GPIO.OUT)


def holdPosition(pulseWidth, holdTime):
    startTime = time.time()

    while (time.time() - startTime) < holdTime:
        GPIO.output(servoPin, GPIO.HIGH)
        time.sleep(pulseWidth / 1_000_000)

        GPIO.output(servoPin, GPIO.LOW)
        time.sleep((20_000 - pulseWidth) / 1_000_000)


def loop():
    # Move servo to 0 degrees and hold for 2 seconds
    holdPosition(500, 2)

    # Move servo to 90 degrees and hold for 2 seconds
    holdPosition(1500, 2)

    # Move servo to 180 degrees and hold for 2 seconds
    holdPosition(2500, 2)


setup()

try:
    while True:
        loop()

except KeyboardInterrupt:
    GPIO.cleanup()
```

```python
import Jetson.GPIO as GPIO
import time

SERVO_PIN = 32   # example — must be a hardware PWM pin on your Jetson

GPIO.setmode(GPIO.BOARD)

pwm = GPIO.PWM(SERVO_PIN, 50)  # 50 Hz = 20 ms period
pwm.start(7.5)                 # ~1500 us

try:
    while True:
        pwm.ChangeDutyCycle(5.0)    # ~1000 us
        time.sleep(2)

        pwm.ChangeDutyCycle(7.5)    # ~1500 us
        time.sleep(2)

        pwm.ChangeDutyCycle(10.0)   # ~2000 us
        time.sleep(2)

finally:
    pwm.stop()
    GPIO.cleanup()
```
